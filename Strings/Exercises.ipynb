```python
# 8.1 string reverse
def reverse_string(s):
    return s[::-1]

# print the reversed string
print(reverse_string('Hello, World!'))
```

```python
# 8.2 replace one space with two spaces
def add_space(word):
    return word.replace(' ', '  ')

# print new word with extra space
print(add_space('Hello Kosmos'))
```

```python
# 8.3 Convert a Phrase to all caps acronym
def phrase_acronym(phrase):
    words = phrase.split(' ')
    acronym = ''
    for word in words:
        acronym += word[0].upper()
    return acronym

# print acronym
print(phrase_acronym('brothers of struggle'))
print(phrase_acronym('Hannah Echols'))
```

```python
# 8.4 singular to plural
def singular_to_plural(word):
    if word.endswith('s'):
        return word + 'es'
    # python or
    # elif word.endswith('f') | word.endswith('fe'):
    elif word.endswith('f'):
        if word[-2] == 'e':
            return word + 's'
        else:
            return word[:-1] + 'ves'
    elif word.endswith('fe'):
        return word[:-2] + 'ves'
    elif word.endswith('y'):
        return word[:-1] + 'ies'
    elif word.endswith('x') | word.endswith('s') | word.endswith('z') | word.endswith('ss') | word.endswith('sh') | word.endswith('ch'):
        return word + 'es'
    else:
        return word +'s'
    
# print plural
print(singular_to_plural('sofa'))
print(singular_to_plural('book'))
print(singular_to_plural('wife'))
print(singular_to_plural('leaf'))
print(singular_to_plural('brief'))

'''
Not a complete program
There are many more rules with singular to pluarl 
e.g. 'child' becomes 'children' and 'foot' becomes 'feet' 
beef is beeves or beefs
This program is a simple demonstration of some rules
'''
```
